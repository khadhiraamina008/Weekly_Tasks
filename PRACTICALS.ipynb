```python
import pandas as pd                                                     # Importing pandas library
# creating dataframe
data = {
    "Department" : ["HR", "IT", "HR", "IT", "Sales", "Sales"],
    "Employee" : ["Amti", "Rahul", "Priya", "Ankit", "sakshi", "Riya"],
    "Salary" : [30000, 50000, 35000, 60000, 45000, 40000],
    "Experience" : [2, 5, 3, 6, 4, 2]
}
df = pd.DataFrame(data)
print("Original DataFrame :")                                           # Displaying original DataFrame
print(df)
grouped = df.groupby("Department")                                      # Grouping based on Department column
# Sum Aggregation
print("\nSum of Salary by Department :")                                # Total salary for each department
print(grouped["Salary"].sum())
# Average Aggregation
print("\nAverage Salary by Department :")                               # Average salary for each department
print(grouped["Salary"].mean())
# Maximum Aggregation
print("\nMaximum Salary by Department :")                               # Max salary in each department
print(grouped["Salary"].max())
# Minimum Aggregation
print("\nMinimum Salary by Department :")                               # Min salary in each department
print(grouped["Salary"].min())
print("\nNumber of Employees in each Department :")                     # Counting employees in each department
print(grouped["Employee"].count())
print("\nMultiple Aggregations on Salary :")                            # Multiple aggregations sum, mean, max, min
print(grouped["Salary"].agg(['sum','mean','max','min']))
print("\nGrouped by Department and Experience :")                       # Grouping using Department and Experience
print(df.groupby(['Department','Experience'])['Salary'].sum())
print("\nIterating through each group:")                                # Iterating through each group
for name, group in df.groupby("Department"):
    print(f"\nDepartment: {name}")
    print(group)
print("\nDataFrame sorted by Salary :")                                 # Sorting DataFrame by Salary column
print(df.sort_values('Salary'))
```

```python
import pandas as pd # Importing pandas library
import numpy as np # Importing numpy library

# Creating sample DataFrame 
data = { 
    "Name": ["Sakshi", "Rahul", "Priya", "Ankit", "Rahul", "Riya", None],
    "Age": [20, 21, np.nan, 22, 21, 20, 23],
    "Marks": [95, np.nan, 90, 85, 88, 92, 80],
    "Department": ["IT", "HR", None, "Sales", "HR", "IT", "HR"]
}

df = pd.DataFrame(data)                              # Creating DataFrame
print("Original Dataset:")                           # Displaying original dataset
print(df)

print("\nMissing Values in Each Column:")            # Checking missing values
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())       # Filling missing numeric values with mean
df["Marks"] = df["Marks"].fillna(df["Marks"].mean()) # Filling missing numeric values with mean

df["Name"] = df["Name"].fillna("Unknown")            # Filling missing categorical with default
df["Department"] = df["Department"].fillna("Unknown")# Filling missing categorical with default

print("\nAfter Filling Missing Values:")             # Displaying after filling missing values
print(df)

df.drop_duplicates(inplace=True)                     # Removing duplicate records
print("\nAfter Removing Duplicates:")                # Displaying after removing duplicates
print(df)

df["Age"] = df["Age"].astype(int)                    # Converting Age column to integer 
print("\nAfter Converting Age to int:")              # Displaying data types
print(df.dtypes)

df.rename(columns={"Marks": "Score"}, inplace=True) # Renaming Marks column to Score
print("\nAfter Renaming Marks to Score:")           # Displaying after renaming
print(df)

df_sorted = df.sort_values("Score", ascending=False)# Sorting in descending order by Score
print("\nDataset Sorted by Score (Descending):")    # Displaying sorted dataset
print(df_sorted)

print("\nRecords Where Score > 85:")                # Filtering records where Score > 85
print(df[df["Score"] > 85])

print("\nFinal Dataset Information:")               # Displaying final dataset info using info()
df.info()
```

```python
import pandas as pd                              # Importing pandas library
# Creating sample dataset containing students marks
data = { 
    "Name": ["Sakshi", "Rahul", "Priya", "Ankit", "Riya", "Aman", "Sneha"],
    "Marks": [95, 88, 90, 85, 88, 92, 80]
}

df = pd.DataFrame(data)                          # Converting dataset into DataFrame
print("Dataset:")                                # Displaying the dataset
print(df)

mean_marks = df["Marks"].mean()                  # Calculating mean of marks
print(f"\nMean of Marks: {mean_marks}")

median_marks = df["Marks"].median()              # Calculating median of marks
print(f"Median of Marks: {median_marks}")

mode_marks = df["Marks"].mode()                  # Calculating mode of marks
print(f"Mode of Marks: {mode_marks.values}")

std_marks = df["Marks"].std()                    # Calculating standard deviation of marks
print(f"Standard Deviation of Marks: {std_marks}")

min_marks = df["Marks"].min()                    # Finding minimum marks
max_marks = df["Marks"].max()                    # Finding maximum marks
print(f"Minimum Marks: {min_marks}")
print(f"Maximum Marks: {max_marks}")

variance_marks = df["Marks"].var()               # Calculating variance of marks
print(f"Variance of Marks: {variance_marks}")

print("\nStatistical Summary using describe():") # Displaying statistical summary
print(df["Marks"].describe())
```

```python
import pandas as pd            # Importing pandas library
import seaborn as sns          # Importing seaborn library
import matplotlib.pyplot as plt # Importing matplotlib library

# ----creating dataset------ 
data={                        # Creating dataset for visualization
    "Maths":[85,90,78,92,88],
    "Science":[80,85,75,98,67],
    "Social":[78,88,72,90,84]
}

# ----creating data frame------
df=pd.DataFrame(data)         # Creating DataFrame
print("==========DATASET============")             # Displaying dataset
print(df)

#-------HEATMAP-----------
plt.figure(figsize=(5,4))     # Creating figure for heatmap
sns.heatmap(df, annot=True, cmap='coolwarm') # Creating heatmap with annotations
plt.title("HEATMAP")          # Adding title
plt.show()                    # Displaying heatmap

#----------BOXPLOT-------------
plt.figure(figsize=(5,4))     # Creating figure for boxplot
sns.boxplot(data=df)          # Creating boxplot
plt.title("BOXPLOT")          # Adding title
plt.show()                    # Displaying boxplot

#---------PAIRPLOT-------------
sns.pairplot(df)              # Creating pairplot
plt.show()                    # Displaying pairplot
```
