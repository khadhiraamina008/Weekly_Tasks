# Generator based CSV Parser

```python
def parse_csv(Filename):
    with open(Filename,"r",buffering=8192) as file:
        next(file)
        for line in file:
            line=line.strip()
            if not line:
                continue
            yield line.split()
            
```

```python
def process_file(filename):
    count = 0
    mean = 0
    M2 = 0
    category_frequency = {}

    for row in parse_csv(filename):
        # Skip rows that don't have at least 3 columns
        if len(row) < 3:
            continue

        transaction_id = row[0].strip()
        category = row[1].strip()

        try:
            amount = float(row[2])
        except ValueError:
            # Skip rows where amount is not a valid number
            continue

        count += 1
        # Welford’s algorithm for mean and variance
        delta = amount - mean
        mean += delta / count
        delta2 = amount - mean
        M2 += delta * delta2

        # Update category frequency
        category_frequency[category] = category_frequency.get(category, 0) + 1

    variance = M2 / (count - 1) if count > 1 else 0
    return count, mean, variance, category_frequency
```

# Execute Aggregation

```python
count, mean, variance, category_frequency = process_file("transaction.csv")
print("Total Records:",count)
print("Running Mean:",round(mean,2))
print("Variance:",round(variance,2))
print("\nCategory Frequencies")
for category, freq in category_frequency.items():
    print(category,":", freq)
    
```
